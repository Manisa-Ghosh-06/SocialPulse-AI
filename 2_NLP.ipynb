```python
import pandas as pd

df = pd.read_csv("cleaned_twitter_data.csv")

df.head()
```

```python
df.columns
```

```python
df.shape
```

**Step 1: Text Cleaning.**

```python
df["TEXT"].head(10)
```

```python
import re

def clean_text(text):
    text = text.lower()                      # lowercase
    text = re.sub(r"http\S+|www\S+", "", text)  # remove URLs
    text = re.sub(r"@\w+", "", text)         # remove mentions
    text = re.sub(r"[^a-zA-Z\s]", "", text)  # remove special characters
    text = re.sub(r"\s+", " ", text).strip() # remove extra spaces
    
    return text
```

```python
df["CLEAN_TEXT"] = df["TEXT"].apply(clean_text)
```

```python
df[["TEXT", "CLEAN_TEXT"]].head(10)
```

**Step 2 — Tokenization.**

```python
import nltk
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')      
```

```python
from nltk.tokenize import word_tokenize
```

```python
df["TOKENS"] = df["CLEAN_TEXT"].apply(word_tokenize)
```

```python
df[["CLEAN_TEXT", "TOKENS"]].head()
```

**Step 3 — Stopword Removal.**

```python
from nltk.corpus import stopwords

nltk.download("stopwords")
```

```python
stop_words = set(stopwords.words("english")) - {"not", "no", "nor"}
```

```python
stop_words
```

```python
df["NO_STOPWORDS"] = df["TOKENS"].apply(
    lambda tokens: [word for word in tokens if word not in stop_words]
)
```

```python
df[["TOKENS", "NO_STOPWORDS"]].head()
```

**Step 4 — Stemming.**

```python
from nltk.stem import PorterStemmer, SnowballStemmer

porter = PorterStemmer()
snowball = SnowballStemmer("english")
```

```python
df["PORTER_STEMMED"] = df["NO_STOPWORDS"].apply(
    lambda tokens: [porter.stem(word) for word in tokens]
)
```

```python
df["SNOWBALL_STEMMED"] = df["NO_STOPWORDS"].apply(
    lambda tokens: [snowball.stem(word) for word in tokens]
)
```

```python
df[["NO_STOPWORDS", "PORTER_STEMMED", "SNOWBALL_STEMMED"]].head(10)
```

```python
df.columns
```

```python
df["PORTER_TEXT"] = df["PORTER_STEMMED"].apply(
    lambda tokens: " ".join(tokens)
)

df["SNOWBALL_TEXT"] = df["SNOWBALL_STEMMED"].apply(
    lambda tokens: " ".join(tokens)
)
```

```python
df[["PORTER_TEXT", "SNOWBALL_TEXT"]].head()
```

**Step 5 — Lemmatization.**

```python
from nltk.stem import WordNetLemmatizer

nltk.download("wordnet")
nltk.download("omw-1.4")
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
df["LEMMATIZED"] = df["NO_STOPWORDS"].apply(
    lambda tokens: [lemmatizer.lemmatize(word) for word in tokens]
)
```

```python
df[["NO_STOPWORDS", "LEMMATIZED"]].head(10)
```

**Step 6 — N-grams**

```python
from nltk.util import ngrams
```

```python
def get_bigrams(tokens):
    return list(ngrams(tokens, 2))
```

```python
df["BIGRAMS"] = df["NO_STOPWORDS"].apply(get_bigrams)
```

```python
df[["NO_STOPWORDS", "BIGRAMS"]].head(10)
```

**EXPERIMENT WHICH PREPROCESSING TO USE with TFIDF**

```python
df["LEMMATIZED_TEXT"] = df["LEMMATIZED"].apply(
    lambda tokens: " ".join(tokens)
)
```

```python
df[[
    "PORTER_TEXT",
    "SNOWBALL_TEXT",
    "LEMMATIZED_TEXT"
]].head(10)
```

```python
X_porter = df["PORTER_TEXT"]
X_snowball = df["SNOWBALL_TEXT"]
X_lemmatized = df["LEMMATIZED_TEXT"]

y = df["SENTIMENT"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df.index,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
porter_train = df.loc[X_train, "PORTER_TEXT"]
porter_test = df.loc[X_test, "PORTER_TEXT"]
```

```python
snowball_train = df.loc[X_train, "SNOWBALL_TEXT"]
snowball_test = df.loc[X_test, "SNOWBALL_TEXT"]
```

```python
lemmatized_train = df.loc[X_train, "LEMMATIZED_TEXT"]
lemmatized_test = df.loc[X_test, "LEMMATIZED_TEXT"]
```

```python
df = df[df["PORTER_TEXT"].str.len() > 0]
df = df.drop_duplicates(subset="PORTER_TEXT")
df = df.reset_index(drop=True)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
ngram_options = {
    "Unigram": (1, 1),
    "Bigram": (2, 2),
    "Unigram + Bigram": (1, 2),
    "Trigram": (3, 3),
    "Bigram + Trigram": (2, 3),
    "Unigram + Bigram + Trigram": (1, 3)
}
```

```python
preprocessing_options = {
    "Porter": (porter_train, porter_test),
    "Snowball": (snowball_train, snowball_test),
    "Lemmatization": (lemmatized_train, lemmatized_test)
}
```

```python
results = []
```

```python
for preprocessing_name, (train_text, test_text) in preprocessing_options.items():

    for ngram_name, ngram_range in ngram_options.items():

        # TF-IDF
        tfidf = TfidfVectorizer(
            ngram_range=ngram_range
        )

        X_train_tfidf = tfidf.fit_transform(train_text)
        X_test_tfidf = tfidf.transform(test_text)

        # Logistic Regression
        model = LogisticRegression(max_iter=1000)

        model.fit(X_train_tfidf, y.loc[X_train])

        # Prediction
        y_pred = model.predict(X_test_tfidf)

        # Evaluation
        accuracy = accuracy_score(y.loc[X_test], y_pred)

        precision = precision_score(
            y.loc[X_test],
            y_pred,
            average="weighted"
        )

        recall = recall_score(
            y.loc[X_test],
            y_pred,
            average="weighted"
        )

        f1 = f1_score(
            y.loc[X_test],
            y_pred,
            average="weighted"
        )

        # Store results
        results.append({
            "Preprocessing": preprocessing_name,
            "N-gram": ngram_name,
            "Accuracy": accuracy,
            "Precision": precision,
            "Recall": recall,
            "F1-Score": f1
        })
```

```python
results_df = pd.DataFrame(results)
results_df
```

```python
results_df.sort_values(
    by="Accuracy",
    ascending=False
)
```

```python
best_result = results_df.loc[
    results_df["Accuracy"].idxmax()
]

print(best_result)
```

```python
results_df[
    ["Preprocessing", "N-gram", "Accuracy"]
].sort_values(
    by="Accuracy",
    ascending=False
)
```

**EXPERIMENT WHICH PREPROCESSING TO USE with BAG OF WORDS**

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
ngram_options = {
    "Unigram": (1, 1),
    "Bigram": (2, 2),
    "Unigram + Bigram": (1, 2),
    "Trigram": (3, 3),
    "Bigram + Trigram": (2, 3),
    "Unigram + Bigram + Trigram": (1, 3)
}
```

```python
preprocessing_options = {
    "Porter": (porter_train, porter_test),
    "Snowball": (snowball_train, snowball_test),
    "Lemmatization": (lemmatized_train, lemmatized_test)
}
```

```python
bow_results = []
```

```python
for preprocessing_name, (train_text, test_text) in preprocessing_options.items():

    for ngram_name, ngram_range in ngram_options.items():

        # Bag of Words
        vectorizer = CountVectorizer(
            ngram_range=ngram_range
        )

        X_train_bow = vectorizer.fit_transform(train_text)
        X_test_bow = vectorizer.transform(test_text)

        # Logistic Regression
        model = LogisticRegression(max_iter=1000)

        model.fit(
            X_train_bow,
            y.loc[X_train]
        )

        # Prediction
        y_pred = model.predict(X_test_bow)

        # Evaluation
        accuracy = accuracy_score(
            y.loc[X_test],
            y_pred
        )

        precision = precision_score(
            y.loc[X_test],
            y_pred,
            average="weighted"
        )

        recall = recall_score(
            y.loc[X_test],
            y_pred,
            average="weighted"
        )

        f1 = f1_score(
            y.loc[X_test],
            y_pred,
            average="weighted"
        )

        # Store results
        bow_results.append({
            "Preprocessing": preprocessing_name,
            "N-gram": ngram_name,
            "Accuracy": accuracy,
            "Precision": precision,
            "Recall": recall,
            "F1-Score": f1
        })
```

```python
bow_results_df = pd.DataFrame(bow_results)
bow_results_df
```

```python
bow_results_df.sort_values(
    by="Accuracy",
    ascending=False
)
```

```python
best_bow_result = (
    bow_results_df
    .sort_values(
        by=["Accuracy", "F1-Score"],
        ascending=False
    )
    .iloc[0]
)

print(best_bow_result)
```

```python
best_tfidf = results_df.loc[
    results_df["Accuracy"].idxmax()
]

best_bow = (
    bow_results_df
    .sort_values(by=["Accuracy", "F1-Score"], ascending=False)
    .iloc[0]
)
```

```python
# Create comparison table

comparison_df = pd.DataFrame([
    {
        "Feature Extraction": "TF-IDF",
        "Preprocessing": best_tfidf["Preprocessing"],
        "N-gram": best_tfidf["N-gram"],
        "Accuracy": best_tfidf["Accuracy"],
        "Precision": best_tfidf["Precision"],
        "Recall": best_tfidf["Recall"],
        "F1-Score": best_tfidf["F1-Score"]
    },
    {
        "Feature Extraction": "Bag of Words",
        "Preprocessing": best_bow["Preprocessing"],
        "N-gram": best_bow["N-gram"],
        "Accuracy": best_bow["Accuracy"],
        "Precision": best_bow["Precision"],
        "Recall": best_bow["Recall"],
        "F1-Score": best_bow["F1-Score"]
    }
])

print(comparison_df)
```

```python
comparison_df[
    ["Accuracy", "Precision", "Recall", "F1-Score"]
] = comparison_df[
    ["Accuracy", "Precision", "Recall", "F1-Score"]
] * 100

comparison_df.round(2)
```
