```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

import warnings

warnings.filterwarnings('ignore')
```

```python
df.dropna(subset=["TEXT", "SENTIMENT"], inplace=True)
df.drop_duplicates(inplace=True)
df = df[df["SENTIMENT"] != "Irrelevant"]
```
